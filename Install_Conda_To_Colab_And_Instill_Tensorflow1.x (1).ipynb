# Install conda environment

TensorFlow 1.x no longer supports higher Python versions. After testing, Python 3.6 is currently the only compatible version. Unfortunately, Google Colab only allows Python 3.7 or higher to be installed. Therefore, to use TensorFlow 1.x, we must install a Conda environment within Google Colab.

```python
!wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O Miniconda3.sh
!ls
!chmod +x Miniconda3.sh
```

```python
%env PYTHONPATH=
!./Miniconda3.sh -b -f -p /usr/local
```
